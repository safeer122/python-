```python
#sort by roll number
students = {}

roll_no = int(input("Enter roll number for student 1: "))
name = input("Enter name for student 1: ")
students[roll_no] = name

roll_no = int(input("Enter roll number for student 2: "))
name = input("Enter name for student 2: ")
students[roll_no] = name

roll_no = int(input("Enter roll number for student 3: "))
name = input("Enter name for student 3: ")
students[roll_no] = name

roll_no = int(input("Enter roll number for student 4: "))
name = input("Enter name for student 4: ")
students[roll_no] = name

roll_no = int(input("Enter roll number for student 5: "))
name = input("Enter name for student 5: ")
students[roll_no] = name

asc_sorted = dict(sorted(students.items()))
desc_sorted = dict(sorted(students.items(), reverse=True))

print("\nOriginal Dictionary:", students)
print("Ascending order by roll no:", asc_sorted)
print("Descending order by roll no:", desc_sorted)
```

```python
# merging two dictionaries
dic1 = {}

roll_no = int(input("Enter roll number for student 1: "))
name = input("Enter name for student 1: ")
dic1[roll_no] = name

roll_no = int(input("Enter roll number for student 2: "))
name = input("Enter name for student 2: ")
dic1[roll_no] = name

roll_no = int(input("Enter roll number for student 3: "))
name = input("Enter name for student 3: ")
dic1[roll_no] = name


dic2 = {}

roll_no = int(input("Enter roll number for student 1: "))
name = input("Enter name for student 1: ")
dic2[roll_no] = name

roll_no = int(input("Enter roll number for student 2: "))
name = input("Enter name for student 2: ")
dic2[roll_no] = name


print("Dictionary 1:", dic1)
print("Dictionary 2:", dic2)

dic1.update(dic2)

print("\nMerged Dictionary:", dic1)
```
