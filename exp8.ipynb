```python
#
name1 = input("Enter name of student 1: ")
roll1 = input("Enter roll number: ")

name2 = input("Enter name of student 2: ")
roll2 = input("Enter roll number: ")

name3 = input("Enter name of student 3: ")
roll3 = input("Enter roll number: ")

name4 = input("Enter name of student 4: ")
roll4 = input("Enter roll number: ")

name5 = input("Enter name of student 5: ")
roll5 = input("Enter roll number: ")

students = {
    roll1: name1,
    roll2: name2,
    roll3: name3,
    roll4: name4,
    roll5: name5
}

ascending = dict(sorted(students.items()))
descending = dict(sorted(students.items(), reverse=True))

print("Original dictionary:", students)
print("Ascending order:", ascending)
print("Descending order:", descending)
```
