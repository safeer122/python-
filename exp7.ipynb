```python
#print the colors contain in color1 not in color2
input1=input("enter colors in set 1:")
input2=input("enter colors in set 2:")

color_set1={color.strip()for color in input1.split(",") }
color_set2={color.strip()for color in input2.split(",") }

print("orginal set 1:",color_set1)
print("orginal set 2:",color_set2)

result = color_set1 - color_set2

print("colors in set 1 but not in set 2:",result)
```

```python
#print all colors from set1&2
input1=input("enter colors in set 1:")
input2=input("enter colors in set 2:")

color_set1={color.strip()for color in input1.split(",") }
color_set2={color.strip()for color in input2.split(",") }

print("orginal set 1:",color_set1)
print("orginal set 2:",color_set2)

result = color_set1 | color_set2

print("all colors:",result)
```
