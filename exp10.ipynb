```python
#count the occurence of each word in a line of text
text = input("Enter a line of text: ")

words = text.split()
count = {}

for word in words:
    count[word] = count.get(word, 0) + 1

for word, frequency in count.items():
    print(word, frequency)
```

```python
#prompt the user for the list of integer for all values grater than 100
numbers = input("Enter integers separated by commas: ").split(",")

result = []

for num in numbers:
    num = int(num)
    if num > 100:
        result.append("over")
    else:
        result.append(num)

print(result)
```

```python
#count the occurence of a within the list
names = input("Enter first names separated by spaces: ").split()

count = 0

for name in names:
    count += name.lower().count("a")

print("The letter 'a' occurs", count, "times")
```

```python
#gcd of two numbers
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

while b != 0:
    a, b = b, a % b

print("GCD =", a)
```

```python
# A list removng even numbers
numbers = input("Enter integers separated by spaces: ").split()

result = []

for num in numbers:
    num = int(num)
    if num % 2 != 0:
        result.append(num)

print(result)
```

```python
#program to find factorial of number
num = int(input("Enter a number: "))

factorial = 1

for i in range(1, num + 1):
    factorial = factorial * i

print("Factorial =", factorial)
```
